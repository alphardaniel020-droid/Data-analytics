```python
Q1=np.percentile(df['age'],25,method='midpoint')
Q2=np.percentile(df['age'],50,method='midpoint')
Q3=np.percentile(df['sg'],50,method='midpoint')
print(Q1)
print(Q2)
print(Q3)
```

```python
df['age'].mean()
```

```python
IQR=Q3-Q2
print(IQR)
```

```python
upperlimit=Q3+1.5*IQR
lowerlimit=Q1-1.5*IQR
print(upperlimit)
print(lowerlimit)
```

```python
outlier=[]
for x in df ['age']:
    if((x>upperlimit),(x<lowerlimit)):
        outlier.append(x)
outlier
```

```python
Q1 = df[colm].quantile(0.25)
Q3 = df[colm].quantile(0.75)
IQR = Q3 - Q1
df_cleaned = df[
    ~(
        ((df[colm] < (Q1 - 1.5 * IQR)) | 
        (df[colm] > (Q3 + 1.5 * IQR)))
        .any(axis=1))]
print("Before:", df.shape)
print("After:", df_cleaned.shape)
```
